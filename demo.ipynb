```python
# inheritance
class parent:
    def show(self):
        print("parent")
class child(parent):
    pass
obj = child()
obj.show()
```

```python
# polymorphism 
class cat:
    def sound(self):
        print("Meow")
class dog:
    def sound(self):
        print("bark")
for animal in [cat(),dog()]:
    animal.sound()
```

```python
# encapsulation
class Bank:
    def __init__(self):
        self.__balance = 1000
    def show_balance(self):
        print(self.__balance)
obj = Bank()
obj.show_balance()
```

```python
# Abstraction
from abc import ABC,abstractmethod
class vehicle(ABC):
    @abstractmethod
    def start(self):
        pass
class car(vehicle):
    def start(self):
        print("car started")
obj = car()
obj.start()
```
