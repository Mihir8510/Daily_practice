```python
# reversed triangle
rows = int(input("Enter the row: "))
for i in range(rows,0,-1): #outer loop for row in reverse
    for j in range(1,i + 1): #inner loop for columns
        print("*", end = " ")
    print() 
```

```python
# pyramid pattern
rows = int(input("enter the row: "))
for i in range(1, rows + 1): #outer loop for rows
    for j in range(rows - i):
        print(" ", end=" ")    #(left side space print thase)
    for k in range(1, 2 * i): #inner loop for starts
        print("*", end = " ")
    print()
```

```python
# inverted pyramid
rows = int(input("enter the rows: "))
for i in range(rows, 0 ,-1):
    for j in range(rows - i):
        print(" ",end = " ")
    for k in range(1, 2 * i):
        print("*", end = " ")
    print()
```

```python
#diamond pattern
rows = int(input("Enter the rows: "))
for i in range(1, rows + 1):
    for j in range(rows - i):
        print(" ",end = " ")
    for k in range(1, 2 * i):
        print("*", end = " ")
    print()                       #(first part pyramind pattern)
for i in range(rows  - 1, 0, -1):
    for j in range(rows - i):
        print(" ",end = " ")
    for k in range(1, 2 * i):
        print("*", end = " ")
    print()                       #(second part inverted pattern)
```

```python
# hollow square pattern 
rows = int(input("Enter the rows: "))
for i in range(1, rows + 1):
    for j in range(1, rows + 1):
        if i == 1 or i == rows or j == 1 or j == rows :
            print("*", end = " ")
        else:
            print(" ",end=" ")
    print()
```

```python
# medium example
#1. ATM withdrawal
balance = 5000
try:
    amount = int(input("Enter amount:"))
    if amount>balance:
        raise Exception("Insufficient balance")
    balance -= amount
    print("Remaining balance:",balance)
except ValueError:
    print("please enter valid amount")
except Exception as e:
    print(e)
```

```python
# 2.login example
try:
    username = input("Enter username:")
    password = input("Enter password")

    if username != "admin" or password != "1234":
        raise Exception("Invalid credentials")
    print("Login successful")
except Exception as e:
    print(e)
```

```python
#3.marks validation example
try:
    marks = int(input("Enter marks:"))
    if marks < 0 or marks > 100:
        raise ValueError("Mark must be between 0 to 100")
    print("valid marks")
except ValueError as e:
    print(e)
```

```python
# 4.bank transfer example
try:
    sender_balnce = 10000
    transfer = int(input("Enter amount:"))

    if transfer > sender_balnce:
        raise Exception("Not enough balance")
    sender_balnce -= transfer

    print("transfer successful")
    print("Remaining balance:",sender_balnce)
except Exception as e:
    print(e)
```
