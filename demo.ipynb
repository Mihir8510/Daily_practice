```python
# pyramid pattern
rows = int(input("enter the row: "))
for i in range(1, rows + 1): #outer loop for rows
    for j in range(rows - i):
        print(" ", end=" ")    #(left side space print thase)
    for k in range(1, 2 * i): #inner loop for starts
        print("*", end = " ")
    print()
```

```python
# inverted pyramid
rows = int(input("enter the rows: "))
for i in range(rows, 0 ,-1):
    for j in range(rows - i):
        print(" ",end = " ")
    for k in range(1, 2 * i):
        print("*", end = " ")
    print()
```

```python
#diamond pattern
rows = int(input("Enter the rows: "))
for i in range(1, rows + 1):
    for j in range(rows - i):
        print(" ",end = " ")
    for k in range(1, 2 * i):
        print("*", end = " ")
    print()                       #(first part pyramind pattern)
for i in range(rows  - 1, 0, -1):
    for j in range(rows - i):
        print(" ",end = " ")
    for k in range(1, 2 * i):
        print("*", end = " ")
    print()                       #(second part inverted pattern)
```

```python
# update (list)
my_list = ["ramesh","faizan","kishan"]
def update_item(old_value,new_value):
    for i in range(len(my_list)):
        if my_list[i] == old_value:
            my_list[i] = new_value
            print(old_value,":",new_value)
update_item("ramesh","mihir")
print(my_list)
```
